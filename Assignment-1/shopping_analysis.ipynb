### 1. Load the CSV Dataset
This code snippet imports the `pandas` and `numpy` libraries, and loads the CSV dataset into a DataFrame.
- **Functions used**: `pd.read_csv()` to read the CSV file.

```python
import pandas as pd
import numpy as np

dataset_path = "Dataset/Combined_dataset.csv"

df = pd.read_csv(dataset_path)

print("Dataset loaded successfully!")
```

### 2. Explore the Dataset
This code snippet explores the structure, metadata, statistics, and dimensions of the loaded DataFrame.
- **Functions/Attributes used**: `.shape` (get rows/cols count), `.columns` (list column names), `.dtypes` (check data types), `.head()` (view first rows), `.tail()` (view last rows), `.info()` (view memory and null information), and `.describe()` (view statistical summaries).

```python
print(f"Dataset contains {df.shape[0]} rows and {df.shape[1]} columns.\n")

print("Column list:")
print(df.columns.tolist())
print("\nData Types:")
print(df.dtypes.head(10))
```

### 3. Handle Missing Values
This code snippet scans the dataset to identify missing values and fills them with appropriate defaults.
- **Functions used**: `.isna().sum()` (detect missing values), and `.fillna()` (replace missing NaN values with specified defaults).

```python
missing_values = df.isna().sum()
print("Columns with missing values:")
print(missing_values[missing_values > 0])
```

```python
#1. Discount (0%)
df['discount'] = df['discount'].fillna(0.0)

#2. Seller name ("Unknown")
df['seller_name'] = df['seller_name'].fillna('Unknown')
df['seller_information'] = df['seller_information'].fillna('Unknown')

print("Remaining missing values:")
print(df.isna().sum()[['discount', 'seller_name', 'seller_information']])
```

### 4. Filter Rows and Select Columns
This code snippet cleans string formats in `final_price` (removing the Rupee symbol and commas) to convert it to a numeric type, adds a mock `quantity` column, and filters rows.
- **Functions/Methods used**: `.str.replace()` and `.str.strip()` (text cleaning), `pd.to_numeric()` (convert data type to numeric), `np.random.randint()` (generate random integers), and boolean indexing (`df[...]`) for filtering.

```python
cleaned_price = df['final_price'].astype(str).str.replace('₹', '', regex=False)
cleaned_price = cleaned_price.str.replace(',', '', regex=False)
cleaned_price = cleaned_price.str.strip('"')

df['price'] = pd.to_numeric(cleaned_price, errors='coerce')

print("Sample of cleaned price:")
print(df[['final_price', 'price']].head())
```

```python
np.random.seed(42)
df['quantity'] = np.random.randint(1, 6, size=len(df))

print("Sample of synthetic quantity column:")
print(df[['product_id', 'title', 'price', 'quantity']].head())
```

```python
high_value_items = df[(df['price'] > 1000) & (df['category'].isin(['backpacks', 'clothing-set', 'casual-shoes']))]
print(f"Found {len(high_value_items)} high value items in selected categories.")

selected_columns = df[['product_id', 'title', 'category', 'price', 'quantity']]
selected_columns.head()
```

### 5. Remove Duplicates
This code snippet duplicates rows to demonstrate duplicate detection and removal.
- **Functions used**: `pd.concat()` (concatenate dataframes), `.duplicated()` (detect duplicate rows), and `.drop_duplicates()` (remove duplicate rows).

```python
df_with_duplicates = pd.concat([df, df.head(5)], ignore_index=True)
print(f"DataFrame size with duplicates: {df_with_duplicates.shape}")

duplicate_rows = df_with_duplicates[df_with_duplicates.duplicated()]
print(f"Number of duplicate rows identified: {len(duplicate_rows)}")

df_cleaned = df_with_duplicates.drop_duplicates()
print(f"DataFrame size after removing duplicates: {df_cleaned.shape}")
```

### 6. Create a Derived Column
This code snippet calculates a new column named `total_amount` as a vector multiplication of the `price` and `quantity` columns.
- **Operators used**: Vectorized multiplication `*` operation across Pandas Series.

```python
df_cleaned['total_amount'] = df_cleaned['price'] * df_cleaned['quantity']

print("Derived column 'total_amount' created successfully!")
df_cleaned[['product_id', 'title', 'price', 'quantity', 'total_amount']].head()
```

### 7. Save the Cleaned Dataset
This code snippet saves the cleaned DataFrame to a new CSV file.
- **Functions used**: `.to_csv()` with `index=False` to save the file.

```python
output_file = "cleaned_shopping_dataset.csv"
df_cleaned.to_csv(output_file, index=False)

print(f"Cleaned dataset saved successfully to '{output_file}'!")
```
